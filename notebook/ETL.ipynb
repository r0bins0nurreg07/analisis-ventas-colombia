```python
    import pandas as pd
```

```python
df= pd.read_csv('../data/raw/sample_superstore.csv', encoding='latin-1 ')    
df.head()

```

```python
#Se transforma las columnas de fecha a formato datetime
df['Order Date']= pd.to_datetime(df['Order Date'])
df['Ship Date']= pd.to_datetime(df['Ship Date'])
df.dtypes
```

```python
#Se crea una nueva columna de margen de ganancia
df['Profit Margin']= (df['Profit']/df['Sales']*100).round(2)
print(df[['Profit', 'Sales', 'Profit Margin']].head())
```

```python
#Se crea la tabla de dimensiones de clientes
dim_customer = df[['Customer ID', 'Customer Name', 'Segment']].drop_duplicates().reset_index(drop=True)
dim_customer.index = range(1, len(dim_customer) + 1)
dim_customer.index.name = 'ID_Customer'
dim_customer = dim_customer.reset_index()
print(dim_customer.head())
dim_customer.shape
```

```python
#Se crea la tabla de dimensiones de productos
dim_product=df[['Product ID','Product Name','Category','Sub-Category']].drop_duplicates().reset_index(drop=True)
dim_product.index=range(1,len(dim_product)+1)
dim_product.index.name='ID_Product'
dim_product=dim_product.reset_index()
print(dim_product.head())
dim_product.shape
                
```

```python
#Se crea la tabla de dimensiones de ubicación
dim_ubication=df[['Country','City','State','Region','Postal Code']].drop_duplicates().reset_index(drop=True)
dim_ubication.index=range(1,len(dim_ubication)+1)
dim_ubication.index.name='ID_Ubication'
dim_ubication=dim_ubication.reset_index()
print(dim_ubication.head()) 
dim_ubication.shape
```

```python
#Se crea la tabla de dimensiones de fecha
dim_date=df[['Order Date']].drop_duplicates().reset_index(drop=True)
dim_date.index=range(1,len(dim_date)+1)
dim_date.index.name='ID_Date'
dim_date=dim_date.reset_index()

dim_date['Year']=dim_date['Order Date'].dt.year
dim_date['Month']=dim_date['Order Date'].dt.month
dim_date['Month_Name']=dim_date['Order Date'].dt.strftime('%B')
dim_date['Quarter']=dim_date['Order Date'].dt.quarter
dim_date['Day_Of_Week']=dim_date['Order Date'].dt.strftime('%A')
print(dim_date.head())
dim_date.shape
df.dtypes
```

```python
#Se crea la tabla de hechos de ventas
fact_sales = df.copy()

fact_sales['ID_Customer'] = df.merge(dim_customer, on='Customer ID', how='left')['ID_Customer']
fact_sales['ID_Product'] = df.merge(dim_product, on='Product ID', how='left')['ID_Product']
fact_sales['ID_Date'] = df.merge(dim_date, on='Order Date', how='left')['ID_Date']
fact_sales['ID_Ubication'] = df.merge(dim_ubication, on=['Country', 'City', 'State', 'Region', 'Postal Code'], how='left')['ID_Ubication']
fact_sales['Profit_Margin'] = (df['Profit'] / df['Sales'] * 100).round(2)

fact_sales = fact_sales[['ID_Customer', 'ID_Product', 'ID_Date', 'ID_Ubication', 'Quantity', 'Sales', 'Discount', 'Profit', 'Profit_Margin']]
print(fact_sales.head())
print(fact_sales.dtypes)
```
